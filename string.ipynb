Reverse a string

```python
s1 = "abb"
reverse = s1[::-1]
print(reverse)
```

Check palindrome string

```python
s1 = "abb"
reverse = s1[::-1]
if(s1 == reverse):
    print("palindrome")
else:
    print("not palindrome")
```

Count vowels

```python
s1 = "Count vowels"
count = 0
for ch in s1:
    if ch in "aeiou":
        count += 1
print(count) 
```

Count consonants

```python
s1 = "Count vowels"
count = 0
for ch in s1:
    if ch.isalpha() and ch not in "aeiou":
        count += 1
print(count) 
```

Count words

```python
s1 = "Count words"
for ch in s1:
    if ch.isalpha():
        count = len(s1)
print(count)
```

Remove spaces

```python
s1 = "Remove spaces"
s2 = ""
for ch in s1:
    if ch != " ":
        s2 = s2 + ch
print(s2)
```

Find duplicate characters

```python
s1 = "Find duplicate characters"
freq = {}
for ch in s1:
    freq[ch] = freq.get(ch,0) + 1

for ch in freq:
    if freq[ch] > 1:
        print(ch)
```

First non-repeated character

```python
s1 = "First non-repeated character"
freq = {}
for ch in s1:
    freq[ch] = freq.get(ch,0) + 1

for ch in freq:
    if freq[ch] == 1:
        print(ch)
        break
```

Check anagram

```python
s1 = "silent"
s2 = "listen"
if sorted(s1) == sorted(s2):
    print("Anagram")
else:
    print("not anagram")
```

```python
s1 = "silent"
s2 = "listen"
freq1 = {}
freq2 = {}
for ch in s1:
    freq1[ch] = freq.get(ch,0) + 1
for ch in s2:
    freq2[ch] = freq.get(ch,0) + 1

if freq1 == freq2:
    print("Anagram")
```

Character frequency

Factorial

```python
a = 10
fact = 1
for i in range(1,a+1):
    fact = fact * i
print(fact)
```

Prime number

```python
a = int(input())
d = []
if (a == 1):
    print("not prime")
else:
    for i in range(1,a+1):
        if (a % i == 0):
            d.append(i)

    if (len(d) == 2):
        print("prime")
    else:
        print("not prime")
```

Armstrong number

```python
n = int(input())
digits = len(str(n))
temp = n
total = 0
while temp > 0 :
    digit = temp % 10
    total = total + digit**digits
    temp = temp // 10

if (n==total):
    print("armstrong")
else:
    print("not armstrong")
```

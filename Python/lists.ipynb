Find maximum

```python
lst = [1,2,3,4,5]
largest = 0
largest = max(lst)
print(largest)
```

```python
lst = [1,2,3,4,5]
largest = lst[0]
for num in lst:
    if num > largest:
        largest = num
print(largest)
```

Find minimum

```python
lst = [1,2,3,4,5]
smallest = 0
smallest = min(lst)
print(smallest)
```

```python
lst = [1,2,3,4,5]
smallest = lst[0]
for num in lst:
    if num < smallest:
        smallest = num
print(smallest)
```

Second largest

```python
lst = [1,2,3,4,5]
largest = lst[0]
second = lst[0]
for num in lst:
    if num > largest:
        second = largest
        largest = num
    elif num > second and num != largest:
        second = num
print(second)
```

Remove duplicates

```python
lst = [1,2,2,3,3,4,4]
new = set(lst)
neww = list(new)
print(neww)
```

```python
lst = [1,2,2,3,3,4,4]
lst1 = []
for num in lst:
    if num not in lst1:
        lst1.append(num)
print(lst1)
```

Merge two lists

```python
lst = [1,2,3]
lst1 = [9,8,7]
lst.extend(lst1)
print(lst)
```

Find missing number

```python
lst = [1,2,4,5]
total = 0
n = int(input())
real = n*(n+1)//2
for num in lst:
    total += num
missing = real - total
print(missing)
```
